```python
import kagglehub
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
path = kagglehub.dataset_download(
    "mohammadamireshraghi/blood-cell-cancer-all-4class"
)

print(path)
```

```python
dataset_path = os.path.join(path, "Blood cell Cancer [ALL]")

print(dataset_path)
```

```python
import matplotlib.pyplot as plt
import cv2

classes = os.listdir(dataset_path)

plt.figure(figsize=(12, 8))

for i, cls in enumerate(classes):

    img_name = os.listdir(os.path.join(dataset_path, cls))[0]

    img_path = os.path.join(dataset_path, cls, img_name)

    image = cv2.imread(img_path)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    plt.subplot(2, 2, i + 1)
    plt.imshow(image)
    plt.title(cls)
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
sizes = []

for cls in classes:

    folder = os.path.join(dataset_path, cls)

    for image in os.listdir(folder):

        img = cv2.imread(os.path.join(folder, image))

        if img is not None:
            sizes.append(img.shape)

print("Unique Image Sizes:")
print(set(sizes))
```

```python
broken = []

for cls in classes:

    folder = os.path.join(dataset_path, cls)

    for image in os.listdir(folder):

        img = cv2.imread(os.path.join(folder, image))

        if img is None:
            broken.append(image)

print("Broken Images:", len(broken))
```

```python
counts = []

for cls in classes:
    counts.append(len(os.listdir(os.path.join(dataset_path, cls))))

plt.figure(figsize=(8,5))

plt.bar(classes, counts)

plt.title("Blood Cancer Dataset Distribution")
plt.xlabel("Classes")
plt.ylabel("Number of Images")

plt.xticks(rotation=15)

plt.show()
```
